## Carga de datos

```python
# from os import *
# archivo
f=open('SAZS_2020_2026.csv', 'rt')
data =f.read()
f.close(  )

print( len( data) , type(data ) )
print( '-'*70)
print (data[:500])
```

```python
# Separación en líneas
data_lines = data.split('\n')
print(len(data_lines), type(data_lines))
print('-'*70)
print(data_lines[0])
print(data_lines[1])
```

```python
# Lectura del header por separado
Headers = data_lines[0].split(',')
print(len(Headers), type(Headers))
print('-'*70)
print(Headers)
```

```python
# Tiempos, uso timestampas que son segundo desde 1970-01-01
#
timeStamp = []
for row in range(1, len(data_lines)):
 timeStamp.append(data_lines[row].split(',')[2])


for item in timeStamp[:10]:
    print(item)
```

```python
# Temperatura y Presión atmosférica
#

temp, press = [], []
for row in range(1, len(data_lines)):
            temp.append(data_lines[row].split(',')[6])
            press.append(data_lines[row].split(',')[10])

for i in range(10):
    print( i,temp[i],press[i] )
```

```python
# Son todos los datos compatibles, tengo los mismos 
# cantidad datos en cada lista?

print(len( timeStamp ),len( temp ),len( press ))
```

## Acondicionamiento de los datos

Vamos a convertir los datos en variables graficables.  
Por ahora son sólo strings de texto.

```python
# Convierto los datos a números,
# hasta ahora son strings


buffer = []

for item in timeStamp:
	buffer.append(float(item.strip()))

timeStamp = buffer

buffer = []

for item in temp:
	buffer.append(float(item))
  
temp = buffer

buffer = []

for item in press:
	buffer.append(float(item))

press = buffer
  
print(len(timeStamp), len(temp), len(press))
```

```python
import matplotlib.pyplot as plt
 
plt.plot(timeStamp, temp)
plt.show()
```
